```python
class student:
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno

    def display(self):
        print(f"Name: {self.name}, Roll No: {self.rollno}")

class graduate(student):
    def __init__(self,age,rollno,name):
        super().init
        self.age=age

s=student("ali",101)
s.display()
 
g1=graduate("Sara", 102,22)
g1.display()
print(g1.age)
```
