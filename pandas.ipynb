## 2. Pandas Series
A Series is a one-dimensional labeled array capable of holding any data type. It can be created from a list, NumPy array, or dictionary.

### 2.1. Creating a Series

```python
import pandas as pd

# From a list
s = pd.Series([10, 20, 30, 40])
print(s)

# From a dictionary
s_dict = pd.Series({'a': 10, 'b': 20, 'c': 30})
print(s_dict)
```

```python
# different datatypes in a series
s = pd.Series([10, 20.0, 30, 40])
print(s)
```

### 2.2. Series Arguments

pd.Series(data=None, index=None, dtype=None, name=None, copy=False, fastpath=False)


```python
s = pd.Series(data=[10, 20, 30, 40], index = [3, 'name', 8, 4])
print (s)
```

```python
s = pd.Series(data=[10, 20, 30, 40], dtype = 'float')
print (s)
```

```python
s = pd.Series(data=[10, 20, 30, 40], name = "testing")
print (s)
```

### 2.3. Series attributes

```python
s = pd.Series(data=[10, 20, 30, 40], index = [3, 'name', 8, 4])
print(s.values)

print(s.index)

print(s.dtype)

print(s.shape)

print(s.size)

print(s.ndim)

print(s.name)

print(s.is_unique)
```

```python
print(s.head(2))

print(s.tail(2))
```

### 2.4. Accessing Series

```python
s = pd.Series([20, 30, 40, 50, 60])

print(s[1])

print(s.loc[1])

print(s.iloc[1])
```

```python
import pandas as pd

s = pd.Series(data = [10, 20], index=[4, 0])
print(s[0])
```

```python
s = pd.Series(data = [20, 30, 40, 50, 60], index = ["s1", "s2", "s3", "s4", "s5"])

print(s[1])

print(s['s2'])

print(s.loc['s2'])

print(s.iloc[1])
```

### 2.5. Slicing Series

```python
# slicing

s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])

print(s[1:4])

print(s.iloc[1:4])

print(s.loc['b':'d'])
# Label-based slicing is inclusive of the end label.
```

```python
print(s.iloc[::2])

print(s.loc['a':'e':2])
```

### 2.6. Combining Series

```python
s1 = pd.Series([1, 2], index=['a', 'b'])
s2 = pd.Series([3, 4], index=['c', 'd'])

print(pd.concat([s1, s2])) #in concat, axis is 0 by default
```

## 3. Pandas DataFrame
A DataFrame is a two-dimensional labeled data structure with columns of potentially different types, similar to a spreadsheet.

### 3.1. Creating a DataFrame

```python
# From a dictionary of lists
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'Salary': [50000, 60000, 70000]}
df = pd.DataFrame(data)
print(df)

# From a list of lists
data = [['Alice', 25, 50000], ['Bob', 30, 60000], ['Charlie', 35, 70000]]
df = pd.DataFrame(data, columns=['Name', 'Age', 'Salary'])
print(df)
```

```python
col1 = pd.Series([1, 2, 3])
col2 = pd.Series(['a', 'b'])
col3 = pd.Series([True, False, True, False])

df = pd.concat([col1, col2, col3], axis=1)
df.columns = ['Numbers', 'Letters', 'Flags']

print(df)
```

```python
row1 = pd.Series([1, 2, 3])
row2 = pd.Series(['a', 'b'])

new = pd.DataFrame([row1, row2])
print(new)
```

## 4. Indexing and Slicing DataFrames
You can access specific rows and columns in a DataFrame using various methods like .loc[], .iloc[], or direct indexing.

### 4.1. Accessing columns

```python
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'Salary': [50000, 60000, 70000]})

print(df['Name'])  # Access 'Name' column
```

### 4.2. Accessing rows

```python
print(df.loc[1])  # Access row with label 1
print(df.iloc[1])  # Access row by position 1
```

### 4.3. Accessing specific elements

```python
print(df.loc[1, 'Salary'])  # Access value in row 1, 'Salary' column
```

### 4.4. Slicing Using Row Index

```python
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'Age': [24, 27, 22, 32, 29],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']}

df = pd.DataFrame(data)

# Slicing rows (position 1 to 3, inclusive of start, exclusive of end)
subset = df.iloc[1:4]
print(subset)
```

### 4.5. Slicing Rows and Columns

```python
# Slice rows 1 to 3 and columns 0 to 1
subset = df.iloc[1:4, 0:2]
print(subset)
```

### 4.6. Slicing Specific Columns

```python
subset = df.loc[:, ['Name', 'City']]
print(subset)
```

### 4.7. Conditional Slicing

```python
subset = df[df['Age'] > 25]
print(subset)
```

## 6. Handling Missing Data

### 6.1. Checking for Missing Data

```python
df_with_na = pd.DataFrame({'Name': ['Alice', 'Bob', None], 'Age': [25, None, 35]})
print(df_with_na.isnull())
```

### 6.2. Drop missing values

```python
df_clean = df_with_na.dropna()
print(df_clean)
```

### 6.3. Fill missing values

```python
df_filled = df_with_na.fillna(0)  # Replace NaN with 0
print(df_filled)
```

```python
df_filled = df_with_na['Age'].fillna(df_with_na['Age'].mean())
print (df_filled)
```

```python
df_not_inplace = df_with_na['Age'].fillna(df_with_na['Age'].mean(), inplace= True)
print (df_not_inplace)
print (df_with_na)
```

## 7. Grouping and Aggregating Data
Pandas supports split-apply-combine operations using groupby(). It groups data and allows for aggregation.

```python
data = {'Name': ['Alice', 'Aaron', 'Bob', 'Charlie'],
        'Age': [25, 35, 30, 35],
        'Salary': [50000, 50000, 60000, 70000]}
df = pd.DataFrame(data)

print (df)
```

```python
# Group by 'Age' and compute the mean salary
df_grouped = df.groupby('Age')['Salary'].mean()
print(df_grouped)
```

```python
df[['Age','Salary']].agg(['min','mean','max'])
```

## 8. Merging and Concatenating DataFrames

### 8.1. Merging DataFrames

```python
import pandas as pd

df1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [25, 30]})
df2 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Salary': [50000, 60000]})
df_merged = pd.merge(df1, df2, on='Name')
print(df_merged)
```

### 8.2. Concatenating DataFrames

```python
df_concat = pd.concat([df1, df2], axis=1)
print(df_concat)
```
