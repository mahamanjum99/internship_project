```python
image = cv2.imread("mnb.PNG")
if image is None:
    print("Image not found!")
    exit()
    
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY) #color spaces 
hsv=cv2.cvtColor(image,cv2.COLOR_BGR2HSV)
rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
gauss=cv2.GaussianBlur(image,(5,5),0) #blurring image  mostly gray image 
ret,thresh=cv2.threshold(gray,100,255,cv2.THRESH_BINARY) # seprated obj and bg 
edges=cv2.Canny(gauss,100,200) # getting borders only 
contours , hierarchy=cv2.findContours(edges,cv2.RETR_EXTERNAL ,cv2.CHAIN_APPROX_SIMPLE) #saving boundary points in the form of shape to know exact shape 
#image modes methods 
draws=cv2.drawContours(image,contours,-1,(0,255,0),2)


def mouse_event(event, x, y, flags, param):

    if event == cv2.EVENT_LBUTTONDOWN:

        pixel = image[y, x]
        print(f"X = {x}, Y = {y}")
        print("Pixel =", pixel)

        cv2.circle(image, (x, y), 10, (0, 0, 255), 2)
        cv2.putText( image, f"({x},{y})",(x + 10, y - 10),cv2.FONT_HERSHEY_SIMPLEX,0.5,(255, 0, 0),2,)
        cv2.imshow("Image", image)

cv2.imshow("Image", image)
cv2.imshow("Gray", gray)
cv2.imshow("HSV", hsv)
cv2.imshow("Edges", edges)
cv2.setMouseCallback("Image", mouse_event)
print("Shape :", image.shape)
print("Size  :", image.size)
print("Data Type :", image.dtype)


while True:
    key = cv2.waitKey(1) & 0xFF

    if key == ord('s'):
        cv2.imwrite("saved_image.png", image)
        print("Image Saved Successfully!")

    elif key == 27:
        break

cv2.destroyAllWindows()
```

```python
#color detection in HSV
image = cv2.imread("mnb.PNG")
if image is None:
    print("Image not found!")
    exit()
    
hsv=cv2.cvtColor(image,cv2.COLOR_BGR2HSV)
lower_red1 = (0, 120, 70)
upper_red1 = (10, 255, 255)

lower_red2 = (170, 120, 70)
upper_red2 = (179, 255, 255)

mask1 = cv2.inRange(hsv, lower_red1, upper_red1)  #checks every pixel and seprates red and other obj
mask2 = cv2.inRange(hsv, lower_red2, upper_red2) 
#red pixels ko white banaya rest all black 

mask = mask1 + mask2
#  or mask = cv2.bitwise_or(mask1, mask2)

result = cv2.bitwise_and(image, image, mask=mask)# ab image se sirf red colour wala object lia 

cv2.imshow("Original", image)
cv2.imshow("Mask", mask)
cv2.imshow("Result", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#shape detection 
#image = cv2.imread("shapes.png")
image = cv2.imread("C:/Users/TECHNIFI/Downloads/shapes.png")
if image is None:
    print("Image not found!")
    exit()
    
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY) #color spaces 
ret,thresh=cv2.threshold(gray,100,255,cv2.THRESH_BINARY) # seprated obj and bg 
#ret, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV)
gauss=cv2.GaussianBlur(gray,(5,5),0) 
edges=cv2.Canny(gauss,100,200) # getting borders only 
contours , hierarchy=cv2.findContours(thresh,cv2.RETR_EXTERNAL ,cv2.CHAIN_APPROX_SIMPLE) #saving boundary points in the form of shape to know exact shape 
#image modes methods 
draws=cv2.drawContours(image,contours,-1,(0,255,0),2)
import oracledb
conn = oracledb.connect(
    user="maham_intern",
    password="InternPass123",
    dsn="localhost:1521/FREEPDB1"
)
cur = conn.cursor()

for cnt in contours:
    #cnt is current contour 
    area=cv2.contourArea(cnt)
    if area>500:
        peri=cv2.arcLength(cnt,True)
        approx=cv2.approxPolyDP(cnt,0.02*peri,True) #only give imp points 
        length=len(approx)
        if length == 3:
            shape = "Triangle"

        elif length == 4:
            x, y, w, h = cv2.boundingRect(approx)
            ratio = w / float(h)

            if 0.95 <= ratio <= 1.05:
                shape = "Square"
            else:
                shape = "Rectangle"

        elif length == 5:
           shape = "Pentagon"

        elif length == 6:
           shape = "Hexagon"

        else:
           shape = "Circle"

        print("area=",area)
        print("perimter=",peri)

        cur.execute(
            "INSERT INTO shape_detection_results (shape_name, area, perimeter) VALUES (:s, :a, :p)",
            {"s": shape, "a": float(area), "p": float(peri)}
        )

        x,y,w,h=cv2.boundingRect(approx)
        cv2.rectangle(image,(x,y),(x+w,y+h),(0,255,0),2)
        cv2.putText(image,shape,(x,y-10),cv2.FONT_HERSHEY_COMPLEX ,0.6,(255,0,0),2)
        
conn.commit()
cur.close()
conn.close()
print("All shapes saved to database.")


cv2.imshow("Image", image)
cv2.imshow("Gray", gray)
cv2.imshow("Edges", edges)
print("Shape :", image.shape)
print("Size  :", image.size)
print("Data Type :", image.dtype)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
